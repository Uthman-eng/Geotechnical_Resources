# Google Earth (KML) / coordinates -> AutoCAD

Step-by-step version of [`kml_to_csv.py`](../kml_to_csv.py) and [`plot_points_to_autocad.py`](../plot_points_to_autocad.py).
Run the cells in order. Edit the **placeholders** marked `# <--`.

AutoCAD steps need Windows + full AutoCAD open with a drawing.

## 1. Read the KML and convert to British National Grid

```python
import sys
sys.path.insert(0, "..")  # use the scripts in the tool folder

from kml_to_csv import kml_to_points_dataframe

KML_FILE = "sample_locations.kml"   # <-- your Google Earth export

points = kml_to_points_dataframe(KML_FILE, target_crs="EPSG:27700")
points
```

## 2. Save to CSV

```python
from pathlib import Path

Path("outputs").mkdir(exist_ok=True)
points.to_csv("outputs/points_from_kml.csv", index=False)
```

## 3. Check the AutoCAD connection

```python
from pyautocad import Autocad

acad = Autocad(create_if_not_exists=False)
doc = acad.ActiveDocument
print("Connected to:", doc.Name)
```

## 4. Plot to AutoCAD

Change the settings on the module before calling it. Set `DRY_RUN = True` to preview without AutoCAD.

```python
import plot_points_to_autocad as plot

plot.INPUT_CSV = "outputs/points_from_kml.csv"   # <-- or your own CSV of points
plot.NAME_COLUMN, plot.X_COLUMN, plot.Y_COLUMN = "Name", "easting", "northing"
plot.INPUT_CRS = "EPSG:27700"

plot.DRAW_CIRCLE = True
plot.DRAW_TEXT = True
plot.INSERT_BLOCK = False          # <-- True if your drawing has a borehole block
plot.BLOCK_NAME = "BOREHOLE-A"     # <-- exact block name
plot.ATTRIBUTE_TAG = "BH_No"       # <-- attribute tag in that block
plot.LAYER_NAME = "GI_LOCATIONS"
plot.DRY_RUN = False

plot.main()
```
