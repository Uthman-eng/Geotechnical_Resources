## Exploratory hole locations (LOCA)

```python
tables["LOCA"]
```

## Strata summary (GEOL)

```python
strata_summary(tables["GEOL"], strata_col="GEOL_GEOL")
```

## SPT results by stratum (ISPT + GEOL)

```python
spt = assign_strata_to_spt(tables["ISPT"], tables["GEOL"])
display(spt_summary(spt))
spt
```

## SPT N vs depth

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 8))
for hole, df in spt.groupby("LOCA_ID"):
    ax.plot(df["ISPT_NVAL"], df["ISPT_TOP"], "o-", label=hole)
ax.set_xlim(left=0)
ax.set_ylim(spt["ISPT_TOP"].max() * 1.05, 0)
ax.set_xlabel("SPT N"); ax.set_ylabel("Depth (m bgl)")
ax.grid(alpha=0.3); ax.legend()
plt.show()
```
