# Summary of AGS Data

```python
from python_ags4 import AGS4

tables, headings = AGS4.AGS4_to_dataframe("ags.ags")
print(tables.keys())
tables["LOCA"]
```

## Location, Geology Strata and SPT information

```python
for group in ["LOCA", "GEOL", "ISPT"]:
    if group in tables:
        print(f"\n===== {group} =====")
        print(tables[group].to_string(index=False))
```

## All Column Key Info

```python
column = tables.keys()

for group in column:
    if group in tables:
        print(f"\n===== {group} =====")
        print(tables[group].to_string(index=False))
```
